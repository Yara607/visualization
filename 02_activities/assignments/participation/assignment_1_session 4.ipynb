```python
#import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy

from PIL import Image
import requests
from io import BytesIO
```

```python
#create x and y2
x=np.array(['Luffy','zoro', 'Nami','usopp','Sanji'])
y2=np.array([170,100,90,120,50])

# Fetch an image from the given URL
response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')

# store the content of downloded image into 'image_file'
image_file = BytesIO(response.content)

#open the image
image = Image.open(image_file)
```

```python
#create a figure with specified size
fig,ax=plt.subplots(figsize=(7,3))
# create a line plot with color red
ax.plot(x,y2,color='red')

# Add a new axis to place an image inside the figure at a specific position and size
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])

# Display the image inside the newly added axis
ax_image.imshow(image)

# # Remove axis lines for a clean image display
ax_image.axis('off')
```
