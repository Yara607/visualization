```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

#set the random seed for reproducibility
np.random.seed(613)

#generate 3 arrayes for x(intiger from 0 to 49) and y1 and y2(50 random intiger between 0 to 100)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
# Create a figure and axis with specified size
fig,ax=plt.subplots(figsize=(5,3))

#create 2 line plot from x, y1 and x,y2
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
# Create a figure and axis with specified size
fig, ax = plt.subplots(figsize=(5, 3))

#Plot y1 against x and label it as "Person 1"
ax.plot(x,y1, label = "Person 1" )

#Plot y2 against x and label it as "Person 2"
ax.plot(x,y2, label = "Person 2" )

#add the legend at the lower right corner
ax.legend(loc='lower right')
fig.show()
```

```python
# Create a figure and axes with a specific size
fig, ax = plt.subplots(figsize=(5,3))

# Set the limits of the axes (X-axis from 0 to 10, Y-axis from 0 to 10)
ax.axis([0, 10, 0, 10])

# Add text at the data coordinate (1, 5) with the coordinate system in terms of data
ax.text(1, 5,". Data:(1, 5)" , transform=ax.transData )

# Add text at the axes coordinate (0.5, 0.1) with the coordinate system in terms of axes
ax.text(0.5, 0.1,". Axes:(0.5, 0.1)", transform=ax.transAxes )

# Add text at the figure coordinate (0.2, 0.2) with the coordinate system in terms of figure 
ax.text(0.2, 0.2,". Figure:(0.2, 0.2)", transform=fig.transFigure )
```

```python
#create a figure and axis with specified size
fig, ax = plt.subplots(figsize=(5, 3))

#plot Y1 against x and label it as person 1
ax.scatter(x,y1, label = "Person 1")

#plot Y1 against x and label it as person 1
ax.scatter(x,y2, label = "Person 2")

# specify the location of  legend to the lower right corner
ax.legend(loc='lower right')

ax.annotate(
    'This is important!',   # The text that will be displayed
    xy=(10, 95),            # The point (10, 95) where the arrow will point
    xytext=(20, 94),        # The location (20, 94) where the text will be displayed
    arrowprops=dict(        # Defines the arrow properties
        facecolor="pink",  # The color of the arrow
        shrink=0.05,         # Shrinks the length of the arrow
        )
)
fig.show()
```
