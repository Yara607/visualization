```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.set_style('whitegrid')
# choose our dataset
sns.lineplot(data=tips, 
x='total_bill', # define our x variable
y='tip') # define our y variable
```

```python
tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip')

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
fig = plt.subplots(figsize=(10, 3))
tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'indigo')

tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
tipgraph = sns.scatterplot(data=tips, x='total_bill',y='tip',
                            style = 'time', hue ='day',
                              palette = ['purple','hotpink','deepskyblue','yellowgreen'])

tipgraph.set(title='Tips vs. Total Bill',xlabel='Total Bill ($)',ylabel='Tip Amount ($)')
```

```python
sns.pairplot(data = tips,hue = 'day')
```

```python
#create a relplot in tips dataset
daysplot = sns.relplot(data=tips, 
                       x="total_bill", #set the total bill as x 
                       y="tip", # set the tip as y
                       hue="sex", #set the different colors base in sex(male/femail)
                       col="day",    # Create separate plots for each day
                       kind="scatter", # kind of plot is scatterplot
                       col_wrap=2)   # Arrange the plots in 2 columns
```

```python
import plotly.graph_objects as go # 'go' is 'graph objects’
x1 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))
graph.update_layout(
title="Pirate Scores",
xaxis_title="Pirates",
yaxis_title="Score")
```

```python
graph=go.Figure()

graph.add_trace(go.Scatter(x=x1,y=y1,
                           mode='markers',
                           marker=dict(size=15,
                          color='hotpink',
                          opacity=1,
                          line=dict(width=5,color='purple')
 )))

graph.update_layout(
    title='intractive pirate plot',
    xaxis_title= 'prates',
    yaxis_title='scores',
    width=700, height=400
)
```

```python
!pip install matplotlib_venn
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(["apple","banana","watermelon"])
B = set(["pumpkin","blueberry","apple","key lime"])
```

```python
diagram = venn2_unweighted([A, B],
set_labels = ('Fruits','Pies'),
set_colors=("blue","red"),
alpha=0.75)
diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))
plt.show()
```
