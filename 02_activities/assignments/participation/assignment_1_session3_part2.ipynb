```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

#set the random seed for reproducibility
np.random.seed(613)

#generate 4 arrayes for  and y1 and 
x1 = np.arange(50) # array x1(intiger from 0 to 49)
y1 = np.random.randint(0, 75,50) # array y1(50 random intiger between 0 to 75)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"]) # array x2(5 categotical value)
y2 = np.array([110, 180, 240, 99, 220]) ## array y2(5 fixed intiger )
```

```python
## Create a figure with two subplots arranged in one row and two columns whid specified size
fig,(ax1, ax2) = plt.subplots(ncols=2,
                              nrows=1,
                              figsize=(7, 3))

## Create a scatter plot on the first subplot (ax1)
ax1.scatter(x1,y1)
# Create a bar chart on the second subplot (ax2)
ax2.bar(x2,y2)
fig.show()
```

```python
#define the font family,color and size
font={'family':'sans-serif', 'color':'green','size':14 }

## Create a figure with two subplots arranged in one row and two columns whid specified size
fig,(ax1, ax2) = plt.subplots(ncols=2,
                              nrows=1,
                              figsize=(7, 3))
#modify the ax1
ax1.scatter(x1,y1,
            marker='D', #change the marker to square
            color='green') #change the color to green
ax1.set_title("marketing view", fontdict=font) # set the title with specified font
ax1.set_ylabel('sale in 1000$') # set the y-axis label
ax1.set_xlabel('advertising cost in $') # set the x-axis label
ax1.xaxis.set_major_locator(plt.MultipleLocator(5)) # Set the major ticks on the x-axis to be placed at intervals of 5 units.

# define the 5 color for bars
bar_color= ['red','blue','orange','green','yellow']

#modify the ax2
ax2.bar(x2,y2, color=bar_color) 
ax2.set_title("sales department performance", color='purple')# set the title with specified color


fig.show()

```

```python
#make the sample data
x = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes
fig, ax = plt.subplots(figsize=(7, 3))
# call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,color = "indigo")
ax.plot(x, y2,color = "red")
```

```python
# Calculate the standard deviation of y2 (used for error bars)
y2_sd = np.std(y2)

# Create a figure and a single subplot with a specified size 
fig, ax = plt.subplots(figsize=(7, 3))

# Plot y2 against x using a red line
ax.plot(x, y2, color = "red")

# Add error bars to the plot
ax.errorbar(x,y2,                               # x and y data points
            yerr = y2_sd,                       # Error value (standard deviation) applied to y-values
            fmt = "none",                       # Do not display markers (only error bars)
            ecolor="indigo",                    # Set the color of the error bars to indigo
            elinewidth=4,                       # Set the width of the error bar lines
            capsize=6,                          # Add caps at the end of error bars with a size of 6
            capthick=4,                         # Set the thickness of the cap lines to 4
            errorevery=2)                       # Display error bars at every 2nd data point
```
