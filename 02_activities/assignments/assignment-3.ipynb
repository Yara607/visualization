```python
# import libraries
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import plotly.graph_objects as go
import requests
```

```python
# Ensures reproducibility
np.random.seed(123) 
```

```python
# Load dataset
df = pd.read_csv(r"G:\DSI\Visualization_Week\visualization\02_activities\assignments\eco_tourism.csv")  
print(df)
```

```python
# Define x, y, and categorical variable 'Impact'
x= df['Visitor Spending']
y= df['Tourism Receipts']
Impact = df['Impact']

# Define colors and marker symbols for different Impact categories
impact_colors = {'Total': 'orange', 'Direct': 'green'}
impact_symbols = {'Total': 'circle', 'Direct': 'square'}

# Create a Plotly figure object
graph=go.Figure()

# Add a trace for the 'Total' impact category
graph.add_trace(go.Scatter(
    x=x [Impact =='Total'],
    y=y [Impact =='Total'], 
    mode='markers', 
    marker=dict(size=15, 
                color=impact_colors['Total'] ,
                symbol= impact_symbols['Total'],
                opacity=0.8), 
    name='Total'
))

# Add a trace for the 'Direct' impact category
graph.add_trace(go.Scatter(
    x=x [Impact =='Direct'],
    y=y [Impact =='Direct'], 
    mode='markers', 
    marker=dict(size=15, 
                color=impact_colors['Direct'] ,
                symbol= impact_symbols['Direct'],
                opacity=0.8), 
    name='Direct'
))


# Customize the layout of the graph
graph.update_layout(
    title=dict(
        text= 'Economic impact by Tourism',
        font= dict(family="Arial", size=18, color="green")
    ),
xaxis= dict(title='Visitor Spending',
            titlefont=dict(family="Verdana", size=14, color="black")),
yaxis= dict(title='Tourism Receipts',
             titlefont=dict(family="Verdana", size=14, color="black")),

width=500, height=500, 
legend=dict(title="Impact")
)
graph.show()
```
